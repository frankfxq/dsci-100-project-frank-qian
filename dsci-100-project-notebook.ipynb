#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

```r
### EDIT THIS CODE CELL ###

tidy_data_2 <- data |>
select(seed, reg_wins_pct, tourney_finish) |>
mutate(top_16_or_better = tourney_finish %in% top_16_or_better) |> 
filter(seed != 0)
#all rows instead of just 6 rows

summary <- tidy_data_2 |> summarize(min_reg_wins_pct = min(reg_wins_pct, na.rm = TRUE), 
                                    max_reg_wins_pct = max(reg_wins_pct, na.rm = TRUE),
                                 mean_reg_wins_pct = mean(reg_wins_pct, na.rm = TRUE))

summary
```
